```python
def args_sum(*args):
    return sum(args)
```

```python
args_sum(10,20,30,40,50,60)
```

```python
def args_example(*args):
    print(args)
```

```python
args_example(1234,312423,34,532)
```

```python
def kwargs_example(**kwargs):
    print(kwargs)
```

```python
kwargs_example(apple=100,banan=150,melon=210) # sözlüge çevirme
```

```python
def kwargs_example_2(**kwargs):
    if "apple" in kwargs:
        print("aapple")
    else:
        print(":(")
```

```python
kwargs_example_2(apple=10,banana=15)
```

```python
kwargs_example_2(melon=25)
```
