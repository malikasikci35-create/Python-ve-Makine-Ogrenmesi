```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):
    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
myCar = Car() # oluşturmaz çünkü soyut bir sınıflama yaptık
```

```python
class Tesla(Car):
    pass
```

```python
class Fruit():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f"{self.name}: {self.calories}"

    def __len__(self):
        return self.calories
```

```python
myMeyve = Fruit("banana",180)
```

```python
len(myMeyve)
```

```python
class Train():

    def __init__(self,name):
        self.name = name

    def __getitem__(self,key):
        if key == "a":
            return self.name
        else:
            return "not found"
        
```

```python
myTrain = Train("myTrain")
```
