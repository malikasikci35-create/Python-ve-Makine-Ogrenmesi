```python
# CONTİNUE-BREAK-PASS # döngü içerisnden çıkartma
```

```python
my_list = [10,20,30,40,50,60,70]
```

```python
print("for loop started")
for num in my_list:
    print(num)
print("for loop ended")    
```

```python
40 in my_list
```

```python
for number in my_list:
    print(number)
    if number == 40:
        print("yes")  # burda kod 40 bulduktan sonrada çalışmaya devam eder bunu istemiyorsak break komutunu kullanmalıyız
```

```python
my_list = [10,20,30,40,50,60,70]
```

```python
for number in my_list:
    print(number)
    if number == 40:
        print("yes")
        break     
```

```python
my_list = [10,20,30,40,50,60,70]
```

```python
for number in my_list:
    if number == 40:
        continue
    print(number)    
```

```python
for num in my_list:  # kodu yazarken yazdıgımız yer hakkında ne yapacagımızı bilmiyorsak pass kullanırız
    pass
```
