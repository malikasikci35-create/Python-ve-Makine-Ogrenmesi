```python
 import pandas as pd

data = pd.read_csv("athlete_events.csv (1).zip", compression="zip")

data.head()
```

```python
# verımızı jupıter ıcını attık buradan cesıltlı analızler yapıcaz
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 

sns.set_style("whitegrid")
sns.scatterplot(x = "Height" , y = "Weight" , data = data )
plt.xlabel("Height of Athletes")
plt.ylabel("Weight of Athletes")
plt.title("Athletes Weight vs Height")
plt.show()
```

```python
sns.set_style("darkgrid")
sns.scatterplot(x = "Height" , y = "Weight" , data = data )
plt.xlabel("Height of Athletes")
plt.ylabel("Weight of Athletes")
plt.title("Athletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.lineplot(x = "Height", y="Weight",hue = "Sex", data  = data)
plt.xlabel("Height of  Athletes")
plt.ylabel("Weight of Athlates")
plt.title(" Athlates Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.displot( x = "Height" , hue = "Sex" , data= data )
plt.ylabel("frequency")
plt.title("Athletes Height Histogram")
plt.show()
```

```python
# cizgi seklinde gostermek icin kde seanborn komutununcok kullaniriz grafikler arasinda bir fark yok ama asagidaki daha anlasilir ve guzel
```

```python
sns.set_style("dark")
sns.displot( x = "Height" , hue = "Sex" , data= data , kind = "kde" )
plt.ylabel("frequency")
plt.title("Athletes Height Histogram")
plt.show()
```

```python
sns.barplot(x = "Medal",y ="Height",hue="Sex" ,data=data )
plt.title("Medals  by Height")
plt.show()
```

```python
sns.catplot(x ="Medal" , y ="Height" , hue = "Sex" , col = "Season", data=data)
plt.show()
```

```python
data = np.array([5,7,9,15,20,22,25,30,32,35,37,40,50,55,60,100])
plt.figure(figsize = (6,5)) # tablonun buyuklugunu kucuklugunu ayarlama
sns.boxplot(y=data)
plt.title(" BOX PLOT")
plt.ylabel("data value")
plt.grid(True)
plt.show()
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python

plt.figure(figsize = (15,8))
plt.subplot(1,2,1)
sns.boxplot(x = "class" , y = "age" , data = df)
plt.title("Age by Class")
plt.xlabel("Class")
plt.ylabel("Age")

plt.subplot(1,2,2)
sns.boxplot(x = "class" , y = "fare", data = df)
plt.title("Class by Fare")
plt.xlabel("Class")
plt.ylabel("Fare")

plt.show()
```
