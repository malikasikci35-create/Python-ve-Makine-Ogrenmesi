```python
df = pd.read_csv("7-cyber_attack_data.csv")
```

```python
# ==============================================================================
# SİBER GÜVENLİK VERİ SETİ: DEĞİŞKEN ADI | İNGİLİZCE | TÜRKÇE
# ==============================================================================

# src_packet_rate 
# -> Source-side packet transmission rate
# -> Kaynak tarafı paket gönderim hızı

# dst_packet_rate 
# -> Destination-side packet reception rate
# -> Hedef tarafı paket alım hızı

# avg_payload_size 
# -> Average size of payload in packets
# -> Paketlerdeki ortalama veri yükü boyutu

# connection_duration 
# -> Duration of the connection (in seconds)
# -> Saniye cinsinden bağlantı süresi

# tcp_flag_count 
# -> Number of TCP flag occurrences
# -> Görülen TCP bayraklarının sayısı

# avg_interarrival_time 
# -> Time between packet arrivals
# -> Paketlerin varış süreleri arasındaki ortalama zaman

# failed_login_attempts 
# -> Number of failed login attempts
# -> Başarısız giriş denemelerinin sayısı

# unusual_port_activity_score 
# -> Score representing unusual port usage
# -> Sıra dışı port kullanımını gösteren skor

# session_entropy 
# -> Entropy of session behavior (for anomaly detection)
# -> Anomali tespiti için oturum davranışının entropisi (karmaşıklığı)

# avg_response_delay 
# -> Average delay in server response (in ms)
# -> Milisaniye (ms) cinsinden ortalama sunucu yanıt gecikmesi

# attack_type 
# -> 0 = Normal, 1 = DDoS, 2 = Port Scan
# -> 0 = Normal, 1 = DDoS, 2 = Port Tarama
```

```python
df.info()
```

```python
df.describe()
```

```python
X = df.drop("attack_type",axis=1)
y = df["attack_type"] 
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score = accuracy_score(y_pred,y_test)
print("score",score)
print(classification_report(y_pred,y_test))
print("consusion matrix : \n",confusion_matrix(y_pred,y_test))
```

```python
penalty = ["l1","l2","elasticnet"]
c_values = [100,10,1,0.1,0.01]
solver = ["newton-cg","lbfgs","liblinear","saq","saga","newton-cholesky"]
```

```python
params = dict(penalty=penalty,C=c_values,solver=solver)
```

```python
params
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
```

```python
cv = StratifiedKFold()
```

```python
grid = GridSearchCV(estimator= model,param_grid = params,cv=cv,scoring="accuracy",n_jobs=-1)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
# ==============================================================================================
# --- ONE VS REST (Bire Karşı Hepsi) ---
# Mantık: Her sınıfı diğerlerinin toplamına karşı tek tek test eder (Sınıf sayısı kadar model kurulur).
# Örnek: "Bu veri Kedi mi, yoksa (Köpek+Kuş) mu?" diye bakar.
# Kullanım: Büyük veri setlerinde hızlıdır ve az bellek harcar; genelde ilk tercih edilen yöntemdir.
# ==============================================================================================
# onevsrestmodel = OneVsRestClassifier(LogisticRegression())


# ==============================================================================================
# --- ONE VS ONE (Bire Karşı Bir) ---
# Mantık: Sınıfları ikişerli eşleştirip tüm olası çiftler için ayrı modeller eğitir.
# Örnek: "Kedi mi-Köpek mi?", "Kedi mi-Kuş mu?", "Köpek mi-Kuş mu?" diye ayrı ayrı yarıştırır.
# Kullanım: Sınıf sayısı az olduğunda etkilidir. SVM gibi hantal algoritmalarla iyi çalışır.
# ==============================================================================================
# onevsonemodel = OneVsOneClassifier(LogisticRegression())


# ==============================================================================================
# --- ARALARINDAKİ TEMEL FARK ---
# 1. Model Sayısı: OvR sınıf sayısı kadar (az) model kurarken, OvO sınıfların kombinasyonu kadar (çok) model kurar.
# 2. Hız: OvR "A mı, değil mi?" diye baktığı için hızlıdır; OvO çiftleri tek tek kapıştırdığı için sınıf sayısı artınca yavaşlar.
# 3. Yaklaşım: OvR genel bir ayrım yaparken, OvO sınıfları birbirleriyle doğrudan kıyaslayarak daha detaylı inceler.
# ==============================================================================================
```

```python
from sklearn.multiclass import OneVsOneClassifier,OneVsRestClassifier
```

```python
onevsonemodel = OneVsOneClassifier(LogisticRegression())
```

```python
onevsrestmodel = OneVsRestClassifier(LogisticRegression())
```

```python
onevsonemodel.fit(X_train,y_train)
y_pred = onevsonemodel.predict(X_test)
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix:",confusion_matrix(y_pred,y_test))
```

```python
onevsrestmodel.fit(X_train,y_train)
y_pred = onevsrestmodel.predict(X_test)
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix:",confusion_matrix(y_pred,y_test))
```
