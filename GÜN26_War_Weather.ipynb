```python
# birbiriyle yüksek korelaasyona sahip olan degişkenlerden birini çıkartıyorum çünkü ikisnide aynı anda kullanırsam
# verimin bozulacagını düşünüyorum
```

```python
def corr_drop(df,threshold):
    corr = df.corr()
    columns_drop = set()
    for i in range(len(corr.columns)):
        for j in range (i):
            if abs(corr.iloc[i,j]) > threshold:
                columns_drop.add(corr.columns[i])
    return columns_drop            
```

```python
# yukarıdaki kod çok önemldiir machine learningde işimizi çok kolaylaştıran bir koddur kesinlikle bilmemzi gerekiyor 
# threshold atayacagmız degere istedgimizden yksek olan verileri çıkartmak için kullandık
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaler = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)
```

```python
# vermizi scaler ettikten soran degerler arasında bir yaklaşım oldugunu görüyorum ilk_grafikteki kadar dagını degil degerler
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
# LİnearRegession modelimiz
```

```python
linear = LinearRegression()
linear.fit(X_train_scaler,y_train)
y_pred = linear.predict(X_test_scaler)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print("mae:",mae)
print("mse:",mse)
print("R2 score:",score)
plt.scatter(y_test,y_pred)
plt.show()
```
