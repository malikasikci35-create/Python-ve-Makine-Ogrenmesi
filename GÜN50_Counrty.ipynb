```python
import math


def plot_all_histograms(df, title_prefix=""):
    num_cols = df.select_dtypes(include=[np.number]).columns
    n_cols = 3
    n_rows = math.ceil(len(num_cols) / n_cols)

    plt.figure(figsize=(5 * n_cols, 4 * n_rows))

    for i, col in enumerate(num_cols, 1):
        plt.subplot(n_rows, n_cols, i)
        sns.histplot(df[col], kde=True, bins=30)
        plt.title(f"{title_prefix}{col}")
        plt.xlabel("")
        plt.ylabel("")

    plt.tight_layout()
    plt.show()
```

```python
plot_all_histograms(df)
```

pca _conponentes vermedigimiz icin otamatik kendisi 8 aldi 

```python
pca.explained_variance_ratio_
```

burada 1. sutunu aldigimizda yuzde 55 acikliyor 1. ve 2. aldigimizda 55 + 13 seklinde acikliyor ve boyle toplanarak gidiyor vermizi 

```python
plt.step(list(range(1,10)),np.cumsum(pca.explained_variance_ratio_))
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.ylabel('varyans')
plt.title('varyans')
plt.show()
```

pca.explained_variance_: Oluşturulan her bir temel bileşenin (component) verideki ne kadarlık varyansı (bilgiyi/özü) temsil ettiğini ham sayısal değerler olarak gösterir

plt.plot(pca.explained_variance_ratio_): Bileşenlerin açıkladığı varyans oranlarını çizgi grafiğe (Scree Plot) dökerek, "Kayıp yaşamadan veriyi kaç bileşene (orijinal sütun yerine kaç boyuta) düşürmeliyim?" sorusunun cevabını (dirsek noktasını) görsel olarak bulmanı sağlar

```python
pca_df2.drop(columns=[3,4,5,6,7,8])
```

```python
from sklearn.cluster import KMeans
```

```python
from sklearn.metrics import  silhouette_score
```

```python
import warnings
warnings.filterwarnings('ignore')

wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(pca_df2)
    wcss.append(kmeans.inertia_)   
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.xlabel('number of cluster')
plt.ylabel('wcss')
plt.show()
```

bence en iyisi 3 olucak class sayisi

```python
model = KMeans(n_clusters=3)
model.fit(pca_df2)
```

```python
labels = model.labels_
```

```python
silhouette_score(pca_df2,labels)
```

```python
df['Class'] = labels 
```

```python
df
```

```python
fif,ax = plt.subplots(nrows=1,ncols=2,figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=df,x='Class',y='child_mort')
plt.title('child_more vs class')

plt.subplot(1,2,2)
sns.boxplot(data=df,x='Class',y='income')
plt.title('income vs class')
plt.show()
```

Class 1, en yüksek çocuk ölüm oranına ve en düşük kişi başı gelire sahip en kırılgan (az gelişmiş) ülke kümesini temsil etmektedir. Class 2 ise en düşük çocuk ölümü ve en yüksek gelir düzeyiyle gelişmiş ülkeleri, Class 0 ise bu iki grup arasında kalan gelişmekte olan ülkeleri göstermektedir
