```python
# ARGS,KWARGS
```

```python
# ARGS -> yukarıda 2 yada 3 elemanlı işlemler yaptırtık ama çok büyük elemanlı işlemler yapacagımız
# zaman args kullana biliriz
```

```python
def args_sum(*args):
    return sum(args)
    
```

```python
args_sum(10,20,30,40,50,60)
```

```python
def args_example(*args):
    print(args)
```

```python
args_example(12,13,14,15,15,15,16,61)
```

```python
# KWARGS -> bizim için bir nevi sözlük oluşturur
```

```python
def kwargs_example(**kwargs):
    print(kwargs)
```

```python
kwargs_example(apple=100,banana=150,melon=200)
```

```python
def kwargs_example2(**kwargs):
    if "apple" in kwargs:
        print("appleeeee")
    else:
        print("noooooooo")
```

```python
kwargs_example2(banana=10,melon=20,apple=80)
```

```python
kwargs_example2(banana=15,melon=90)
```
