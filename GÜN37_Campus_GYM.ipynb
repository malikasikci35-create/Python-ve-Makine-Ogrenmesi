```python
X = df.drop('number_people',axis=1)
y = df['number_people']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso 
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
def model_matrix(true,predicted):
    mae = mean_absolute_error(true,predicted)
    mse = mean_squared_error(true,predicted)
    rmse = np.sqrt(mean_squared_error(true,predicted))
    r2_squared  = r2_score(true,predicted)
    return mae,mse,r2_squared
```

```python
models = {
    'Linear Regression':LinearRegression(),
    'Lasso':Lasso(),
    'Ridege': Ridge(),
    'K-Neighbours Regressor':KNeighborsRegressor(),
    'Decision Tree': DecisionTreeRegressor(),
    'Random Forest  Regressor': RandomForestRegressor()
    
}
```

```python
for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train)
    
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    model_train_mae, model_train_rmse, model_train_r2 = model_matrix(y_train, y_train_pred)
    model_test_mae, model_test_rmse, model_test_r2 = model_matrix(y_test, y_test_pred)
    
    print(list(models.values())[i])
    
    print("Evaluation for Training Set")
    print("RMSE :", model_train_rmse)
    print("Mean Absolute Error :", model_train_mae)
    print("R2 Score :", model_train_r2)
    
    print("---------------------")
    
    print("Evaluation for Test Set")
    print("RMSE :", model_test_rmse)
    print("Mean Absolute Error :", model_test_mae)
    print("R2 Score :", model_test_r2)
    
    print("---------------------\n")
```

```python
# 1. Elimizdeki tüm makine öğrenmesi modellerini 'for' döngüsüyle sırayla çağırdık.
# 2. X_train ve y_train ile modelleri eğittik (kalıpları/kuralları öğrenmelerini sağladık).
# 3. Eğitilen modellerle hem Eğitim (Train) hem de Test setleri için tahminler ürettik.
# 4. Yazdığımız metrik fonksiyonunu kullanarak her modelin başarı durumunu hesapladık.

# * MAE & RMSE : Modelin yaptığı tahmin hatalarının büyüklüğünü gösterir.
#                 Bu değerlerin olabildiğince DÜŞÜK (0'a yakın) olmasını isteriz.
#
# * R2 Score : Modelin veriyi açıklama başarısıdır (Doğruluk oranı gibi düşünülebilir).
#                Bu değerin olabildiğince YÜKSEK (1'e yakın) olmasını isteriz.
#
# * Overfittin Kontrolü: 
#   Train setindeki R2 çok yüksek, Test setindeki R2 çok düşükse model ezberlemiştir.
#   Hedefimiz; hem Train hem Test setinde birbirine yakın ve yüksek başarı yakalamaktır.
# ==============================================================================
```
