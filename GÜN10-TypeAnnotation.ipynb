```python
def add_number(a: int,b: int) -> int :
    return a + b                      # iki kod arasında bir fark yoktur sadece type ları belirtilmiştir
```

```python
def process_value(value: int | str) -> str:
    if isinstance(value , int):
        return f"processed integer: {value}"
    else:
        return f"processde string: {value}"  # isinstance komutu şu veri ye ait mi anlamında kullanılır
```

```python
process_value(14)
```

```python
process_value("ali")
```

```python
process_value([10,20,30])
```

```python
from typing import List

def sum_list(numbers: List[int]) -> int:
    return sum(numbers)
```

```python
numbers = [2,3,4,5,6]
sum_list(numbers)
```

```python
numbers = [2,3,4,5,6]
total = sum_list(numbers)
```
