```python
### 📊 Normalizasyon vs. Standardizasyon Özet Tablosu

| Özellik | Normalizasyon (Min-Max Scaling) | Standardizasyon (Z-Score Scaling) |
| :--- | :--- | :--- |
| **Matematiksel Formül** | $$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$|$$X_{std} = \frac{X - \mu}{\sigma}$$ |
| **Ölçek Aralığı** | Tam olarak **[0, 1]** veya **[-1, 1]** arasına sıkıştırır. | Sabit bir sınırı yoktur. Değerlerin çoğu **[-3, 3]** arasında dağılır. |
| **Kritik Değişim** | Verinin ortalaması ve standart sapması değişken kalır. | Verinin **ortalamasını 0, standart sapmasını 1** yapar. |
| **Aykırı Değer Etkisi** | Aykırı değerlerden (outliers) **çok kötü etkilenir**, normal veriyi sıkıştırır. | Aykırı değerlere karşı **daha dirençlidir** ve onları dışlamaz. |
| **En İyi Gittiği Algoritmalar** | KNN, Yapay Sinir Ağları (YSA), Görüntü İşleme. | Lineer/Lojistik Regresyon, PCA, SVM. |
| **Ne Zaman Seçilmeli?** | Verinin dağılımı bilinmiyorsa veya sınır netse. | Veri normal dağılıma yakınsa veya aykırı değer varsa. |
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mse = mean_squared_error(y_test,y_pred_test)
mae = mean_absolute_error(y_test,y_pred_test)
rmse = np.sqrt(mse)
print("mse",mse)
print("mar",mae)
print("rmse",rmse)
```

```python
r2 = r2_score(y_test,y_pred_test)
print("r2 score:",r2)
```
