```python
# ZİP ( örnegin 3 ayrı listeyi birleştirmek istersek zip komuutunu kullnaırız)
```

```python
food_list = ["apple","banana","melon"]
```

```python
calories_list = [100,250,200]
```

```python
day_list = ["monday","tuesday","wednesday"]
```

```python
zipped_list = zip(food_list,calories_list,day_list)
```

```python
zipped_list
```

```python
type(zipped_list)
```

```python
# zip oldugu zaman işimize yaramaz bundan dolayı bunuda listeye çevirmemiz gerekir
```

```python
zipped_list = list(zip(food_list,calories_list,day_list))
```

```python
zipped_list
```

```python
type(zipped_list)
```

```python
len(zipped_list)
```
