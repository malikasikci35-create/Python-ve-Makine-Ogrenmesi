```python
### 1. Karar Ağaçları (Decision Trees)
* **Mantık:** Veriyi yukarıdan aşağıya doğru sorular (koşullar) sorarak dallara ayıran denetimli makine öğrenmesi algoritması.
* **Kullanım:** Hem Sınıflandırma (Classification) hem de Regresyon (Regression) problemlerinde kullanılır.
* **Yapı:** Root Node (Kök) -> Decision Nodes (Dallar) -> Leaf Nodes (Yapraklar/Tahminler).
```

```python
### 2. Gini Impurity vs. Entropy (Saflık Ölçütleri)
Ağacın veriyi nereden böleceğini belirleyen matematiksel ölçütlerdir (Belirsizlik/Düzensizlik ölçer).

* **Gini Impurity:** Rastgele seçilen bir verinin yanlış sınıflandırılma olasılığı. 
  * `0` = Mükemmel saf düğüm (Sıfır hata).
  * Scikit-learn için varsayılandır (Hızlı hesaplanır).
* **Entropy:** Düzensizlik/kaos miktarı. 
  * `0` = Tam düzen (Saf), `1` = Tam kaos (Yarı yarıya karışık).
  * Logaritmik hesaplama yaptığı için Gini'ye göre biraz daha yavaştır.
* **Information Gain (Bilgi Kazancı):** Veri bölündüğünde entropideki düşüş miktarı. Ağaç, kazancı maksimum yapacak bölmeyi seçer.
```

```python
### 3. Pruning (Budama) & Regresyon Mantığı
* **Pruning:** Ağacın veriyi ezberlemesini (**Overfitting**) engellemek için gereksiz derin dalların kesilmesi/sınırlandırılması işlemidir (`max_depth`, `min_samples_leaf`).
* **Decision Tree Regression Mantığı:** 
  * Veri uzayını bölgelere ayırır.
  * Bir yaprağın içine düşen verilerin **ortalama değerini (mean)** tahmin olarak döndürür.
  * Bölme kriteri olarak Gini/Entropy yerine **MSE** (Mean Squared Error) veya **MAE** kullanır.
```

```python
### Hangisi Hangisi? (Hızlı Özet)

*   **Gini Impurity:** **Hız** istiyorsan kullanılır. Logaritma hesabı yapmadığı için büyük veri setlerinde işlem süresini kısaltır (Scikit-learn doğrudan bunu kullanır).
*   **Entropy:** **Hassasiyet** istiyorsan kullanılır. Grupların birbirine olan dengesini ve bilgi kazancını logaritmik formülle daha ince hesaplar ama bilgisayarı biraz daha yorar.

**Özetle:** Büyük veride zaman kazanmak için **Gini**, daha detaylı ve dengeli bir bölünme ölçümü için **Entropy** tercih edilir.
```
