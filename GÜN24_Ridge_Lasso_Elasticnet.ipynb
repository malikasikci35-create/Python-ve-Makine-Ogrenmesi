```python
# Çapraz Doğrulama (Cross Validation - CV) Rehberi

**Çapraz Doğrulama (Cross Validation)**, makine öğrenmesi modellerinin performansını ölçen bir yöntemdir.
Modelin genelleme yeteneğini (**generalization capability**), yani hiç görmediği verideki başarısını test eder.
Model performansını çok daha güvenilir ve sağlam (**robust**) bir şekilde değerlendirmemizi sağlar.

---

### ❓ Neden Kullanırız? (Why Do We Use It?)

* **Ezberlemeyi (Overfitting) Önlemek:** Modelin veriyi sadece ezberleyip ezberlemediğini tespit eder.
* **Veriyi Maksimum Verimle Kullanmak:** Geleneksel yöntemde (**train-test split**) test verisi eğitimde hiç kullanılmaz.
  Bu yöntemde ise her veri parçası hem eğitimde (**training**) hem de testte (**testing**) yer alır.
* **Şans Faktörünü Ortadan Kaldırmak:** Veriyi tek sefer böldüğümüzde, test setine şans eseri kolay veriler denk gelebilir.
  Bu yöntem veriyi farklı kombinasyonlarda bölerek bu varyans riskini sıfırlar.

---

### ⏳ Ne Zaman Kullanırız? (When Do We Use It?)

* Modelin gerçek dünya performansından (**production performance**) emin olmak istediğimizde,
* Veri setimiz küçük olduğunda ve her verinin kritik önem taşıdığı durumlarda,
* Modelin en iyi çalışan ayarlarını, yani hiperparametrelerini (**hyperparameter tuning**) seçerken kullanılır.

---

# Çapraz Doğrulama Yöntemleri ve Çeşitleri

### 1. Leave-P-Out Cross Validation (LPOOCV) - Örn: Leave-2-Out
* **Mantığı:** Veri setinden her seferinde $P$ adet gözlem doğrulama (**validation**) için dışarıda bırakılır.
  Geri kalan tüm verilerle model eğitilir. Bu işlem tüm olası kombinasyonlar için tekrarlanır.
* **Avantajı:** Daha fazla veriyle test yapıldığı için güvenilirlik (**reliability**) çok yüksek olur.
* **Dezavantajı:** Kombinasyon sayısı aşırı arttığı için hesaplama maliyeti (**computational cost**) çok yüksektir.

---

### 2. K-Fold Cross Validation
* **Mantığı:** Veri seti $k$ adet eşit parçaya (**folds / blocks**) bölünür. 
  Her seferinde bir parça doğrulama, kalanlar eğitim için kullanılır. İşlem $k$ kez tekrarlanır.
* **Avantajı:** Dengeli doğrulama (**balanced validation**) sağlar.
* **Avantajı:** Hesaplama maliyeti LOOCV (Leave-One-Out) yöntemine göre oldukça düşüktür.

---

### 3. Stratified K-Fold Cross Validation (Katmanlı K-Fold)
* **Mantığı:** K-Fold yönteminin sınıflandırma (**classification**) problemleri için iyileştirilmiş versiyonudur.
  Her katmandaki hedef etiket oranlarını (örneğin %60 olumlu, %40 olumsuz) sabit tutarak korumaya çalışır.
* **Avantajı:** Dengesiz veri setlerinde (**imbalanced datasets**), doğrulama setinin sınıf dağılımını korur.
  Böylece çok daha doğru ve adil bir değerlendirme (**accurate evaluation**) yapılmasını sağlar.

---

### 4. Time Series Cross Validation (Zaman Serisi Çapraz Doğrulama)
* **Mantığı:** Zaman sırasına bağlı verilerde (borsa, hava durumu vb.) geçmiş verilerle gelecek tahmin edilir.
  Rastgele bölme yapılmaz; verinin kronolojik sırası ve zamansal yapısı (**temporal structure**) sıkı sıkıya korunur.
```
