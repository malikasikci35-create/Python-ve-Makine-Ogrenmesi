```python
sns.set_style("whitegrid")
sns.scatterplot(x = "alcohol" , y="density", data = df)
plt.title("alcol and density (Scatterplot)")
plt.xlabel("ALCOHOL")
plt.ylabel("DENSITIY")
plt.show()
```

```python
sns.set_style("dark")
sns.lineplot(x = "alcohol" , y = "density" , hue = "citric acid" ,  data = df )
plt.title(" Alcohol and Density (Lineplot)")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()
```

```python
# 📊 Grafik Seçim Rehberi (Özet Kopya Kağıdı)

---

### 1. Veri Türüne Göre Grafik Seçimi

| Grafik Türü | Hangi Veri İçin? | Ne Zaman Kullanılır? | Örnek (Şarap Verisi) |
| :--- | :--- | :--- | :--- |
| **`sns.countplot`** | 1 Tane Kategorik | Sınıfların **kaç tane (adet)** olduğunu saymak için | Puanların (`quality`) veride kaçar tane olduğunu görmek. |
| **`plt.pie`** (Pasta) | 1 Tane Kategorik | Parçaların bütüne göre **yüzdelik (%) oranını** görmek için | Şarapların yüzde kaçı 5 veya 8 puan almış bakmak. |
| **`sns.boxplot`** | Kategorik + Sayısal | Grupların **ortalamasını, yayılımını ve aykırı değerlerini** kıyaslamak için | Kaliteye göre alkol (`alcohol`) dağılımını görmek. |
| **`plt.scatter`** | 2 Tane Sayısal | İki sayısal veri arasındaki **ilişkiyi (korelasyonu)** noktalarla görmek için | Alkol oranı ile pH arasındaki bağlantıyı incelemek. |
| **`sns.displot`** | 1 Tane Sayısal | Sayısal verinin **yoğunlaştığı aralıkları (eğriyi)** görmek için | Şarapların genel alkol dağılım haritasını çıkarmak. |

---

### 💡 Analist Altın Kuralları

* ❌ **`plt.plot()` Tuzağı:** Zamana bağlı olmayan yoğun verilerde çizgi grafik kullanma, çizgiler birbirine girer. Her zaman **`plt.scatter()`** kullan.
* ❌ **Sürekli Sayı Tuzağı:** Pasta (`pie`) ve sütun (`barplot`) grafiklerinde `hue` veya eksen kısmına ondalıklı sayı içeren (`density` gibi) sütunları koyma. Sadece gruplanabilir (`quality` gibi) verileri koy.
* **`data=df` Şartı:** Seaborn kullanırken eksen isimlerini tırnak içinde yazıyorsan, fonksiyonun sonuna mutlaka `data=df` ekle.
```

```python

# 1. Sütun isimlerini listele
columns = df.columns

# 2. 4 satır ve 4 sütundan oluşan 16'lık bir grafik matrisi oluştur
fig, ax = plt.subplots(4, 4, figsize=(16, 15))

# 3. Matrisi tek boyutlu bir listeye dönüştür (sırayla erişebilmek için)
ax = ax.flatten()

# 4. Her bir sütun için döngüyü başlat ve grafiklerini çizdir
for i, column in enumerate(columns):
    sns.kdeplot(
        data=df,
        x=column,
        hue=df.quality,  # Buraya hedef değişkenini (sınıflandırma kolonunu) yazabilirsin
        ax=ax[i]
    )
    
    # Grafik başlığını ve X ekseni etiketini düzenle
    ax[i].set_title(f"{column} Distribution")
    ax[i].set_xlabel(None)

# 5. Eğer veri setindeki sütun sayısı 16'dan azsa, boş kalan grafik alanlarını gizle
for i in range(i + 1, len(ax)):
    ax[i].axis("off")

# Grafiklerin birbirinin üstüne binmesini engelle ve ekrana bas
plt.tight_layout()
plt.show()
```
