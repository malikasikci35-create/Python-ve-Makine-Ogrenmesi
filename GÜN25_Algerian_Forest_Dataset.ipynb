```python
# tüm korelason degerlermizi çekmiş olduk  bir for döngüsü ekleyerek birbiriyle çok baglantılı olanlarında 1 ni çıkartalım
```

```python
def correlation_for_dropping(df,threshold):# threshold örnegin 85 yukarısnı çıkart için kullanıyoruz
    corr = df.corr()
    columns_to_drop = set()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop            

    ## çok öneml bir def fonkisyonu 
```

```python
correlation_for_dropping(X_train,0.85)
```

```python
correlation_for_dropping(X_train,0.95)
```

```python
correlation_for_dropping(X_train,0.75)
```

```python
columns_dropping = correlation_for_dropping(X_train,0.85)
```

```python
### Lasso vs LassoCV: Temel Özellikler ve Farklar

*   **Lasso (Standart):** 
    *   $\alpha$ (ceza) parametresini manuel olarak (`alpha=1.0` gibi) sizin vermenizi bekler.
    *   En iyi hiperparametreyi bulmak için harici bir döngü ya da `GridSearchCV` gerekir.
    *   Sadece tek bir model eğittiği için çok hızlıdır.

*   **LassoCV (Cross-Validated):**
    *   Veriyi otomatik olarak parçalara böler (Çapraz Doğrulama / Cross-Validation).
    *   Farklı $\alpha$ değerlerini arka planda dener ve **en ideal $\alpha$ değerini kendisi seçer**.
    *   Doğru parametreyi bulma güvencesi verir ancak işlem yükü daha fazladır.
```
