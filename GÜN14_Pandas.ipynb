```python
# bir kolan daha ekleyelim ve maasi 50000 den dusuk 50.000 ile 80.000 arasinda ise ortalama ve 80.000 den yukse yuksek maas seklinde yeni bir kolan ekleyelim tablomuza
```

```python
def salary_category(salary):
    if salary < 50000:
        return"low maas"
    elif 50000 <= salary < 80000:
        return"ortalama maas"
    else:
        return"yuksek maas"
        
```

```python
salary_category
```

```python
df["salary category"] = df["Salary"].apply(salary_category) # seklinde tablomuza yeni bir veri eklemis olduk
```

```python
df # salart kategorisi
```

```python
# yeni bir performasn skoru olustur calisanin tecrubesi 10 yildan fazla ise performence skoru 1 yil arttir
```

```python
def adjust_performance(experince):
    if experince > 10 :
        return 1 
    else:
        return 0
    
```

```python
df["Adjusted"] = df["Experience"].apply(adjust_performance)
```

```python
df
```

```python
df["newscore"] = df["Performance_Score"] + df["Adjusted"]
```

```python
df
```

```python
# 2. cozum yolu
```

```python
def  adjust_new(row):
    if row["Experience"] > 10:
        return row["Performance_Score"] + 1 
    else:
        return row["Performance_Score"]
```

```python
df["Adjusted_Score"] = df.apply(adjust_new,axis=1)
```
