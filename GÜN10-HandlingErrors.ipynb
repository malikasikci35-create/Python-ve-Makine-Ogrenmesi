```python
try:
    myAge = int(input("enter age:"))
    print(myAge * 2)
except:
    print("enter you age !!!!")
```

```python
# kullanıcı koda verileri dogru girene kadar devam etmesini istiyorsak aşagıdaki gibi yazmalıyız kodu
```

```python
while True:
    try:
        myAge = int(input(" enter age:"))
        print(myAge*2)
        break
    except:
        print("enter you age!!!!!!") # break komutunu koymasak sonsuza kadar bize soru sorar sayı girildigi zaman durmanasını istersek break koyarız
```

```python
# çok kullanılmayan bir yöntem daha var else - finally komutları
```

```python
while True:
    try:
        myAge = int(input(" enter age:"))
        print(myAge*2)
    except:
        print("enter you age!!!!!!") 
    else:
        print(" else excuted")
        break
    finally:
        print("finally") # else komutunu bir nevi break koymak için kullanırız 
                        # finally komutuda eger kodun sonunda illaki çalışmasını istedigimiz bir kod veya yazı varsa ekleriz
```
