```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("6-bank_customers.csv")
```

```python
X = df.drop("subscribed",axis=1)
y = df["subscribed"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state= 15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic = LogisticRegression()
logistic.fit(X_train,y_train)
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
# müşterilerin bankaya üye olucak mı olmuyacak mı onu görüyoruz burda 1 = olucak,, 0 = olmayacak
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
# HYPERPARAMETER TUNİNG 
## (Hiperparametre Ayarlama)

### 📌 Ne Yapacağız?
# Modelimizin performansını (yukarıdaki Confusion Matrix sonuçlarını) en iyi seviyeye getirmek için modelin temel ayarlarını (parametrelerini) optimize edeceğiz.

### 🎯 Neden Yapıyoruz?
# Mevcut hata sayılarını (13 ve 11) daha da düşürmek ve modelden maksimum tahmini doğruluğunu (en net sonucu) almak için yapıyoruz.
```

```python
model = LogisticRegression()
```

```python
penalty = ["l1","l2","elasticnet"]
c_values = [100,10,1,0.1,0.01]
solver = ["newton-cg","lbfgs","liblinear","saq","saga","newton-cholesky"]
```

```python
# ==============================================================================
# 1. PENALTY (Fren Türü) -> Aşırı öğrenmeyi (overfitting) engelleme yöntemi.
# ==============================================================================
# - 'l2' (Ridge): Katsayıları küçültür ama yok etmez (Varsayılan).
# - 'l1' (Lasso): Önemsiz özellikleri tamamen sıfırlar (Değişken eler).
# - 'elasticnet': Hem 'l1' hem 'l2'yi AYNI ANDA karıştırarak kullanır (En dengelisidir).

# ==============================================================================
# 2. C PARAMETRESİ (Frenin Gevşekliği) -> Modelin esneklik ayarı.
# ==============================================================================
# - Küçük C: Fren serttir. Model basit kalır, ezber yapmaz.
# - Büyük C: Fren gevşektir. Model veriye aşırı odaklanır (Ezber riski!).

# ==============================================================================
# 3. SOLVER (Matematik Motoru) -> En doğru katsayıları bulan algoritma.
# ==============================================================================
# - 'lbfgs': Standart ve hızlı motor (Sadece 'l2' destekler).
# - 'liblinear': Küçük veriler için ideal (Sadece 'l1' ve 'l2' destekler).
# - 'saga': En gelişmişidir. 'elasticnet' dahil TÜM ceza türlerini destekler.
```

```python
params = dict(penalty=penalty,C=c_values,solver=solver)
```

```python
params
```

```python
# GRİD SEARCH CV
```

```python
# ==============================================================================
# GRİD SEARCH CV NEDİR? (En Kısa Özet)
# ==============================================================================

# NE İŞE YARAR? -> Modelin için EN İYİ AYARLARI (hiperparametreleri) otomatik bulmaya yarar.
# NEDEN KULLANILIR? -> Tüm kombinasyonları tek tek deneyip veriyi katlayarak (CV) test ettiği için, 
#                    en yüksek doğruluğa sahip güvenilir modeli şansa bırakmadan seçmek amacıyla kullanılır.

# ==============================================================================
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
```

```python
cv = StratifiedKFold()
```

```python
grid = GridSearchCV(estimator= model,param_grid = params,cv=cv,scoring="accuracy",n_jobs=-1) # n_job = -1 tüm cpu kullan demek oluyor
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
# RANDOM SEARCH CV
```

```python
# ==============================================================================
# RANDOMIZED SEARCH CV NEDİR? (En Kısa Özet)
# ==============================================================================

# NE İŞE YARAR? -> Modelin için en iyi ayarları, tüm kombinasyonları değil, 
#                  rastgele seçtiği parametreleri deneyerek çok daha hızlı bulmaya yarar.
#
# NEDEN KULLANILIR? -> GridSearch gibi her şeyi tek tek deneyip saatlerce beklemek yerine,
#                    büyük veri ve geniş parametrelerde çok daha az zaman ve işlemci 
#                    gücü harcayarak en iyiye en yakın sonuçları yakalamak için kullanılır.

# ==============================================================================
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
```

```python
randomcv = RandomizedSearchCV(estimator= model , param_distributions=params,cv=5,n_iter=10,scoring="accuracy")
```

```python
randomcv.fit(X_train,y_train)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("score:",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
# ==============================================================================
# JUPYTER İÇİN MODEL DEĞERLENDİRME VE TABLO YORUMLAMA REHBERİ
# ==============================================================================

# 1. SCORE / ACCURACY (Genel Doğruluk Oranı: 0.92)
# -> Modelin 300 veriden toplamda %92'sini doğru tahmin etmiş. Genel başarıyı gösterir.

# 2. PRECISION (Kesinlik / İsabet Oranı) -> "Tahmin ettiklerimin ne kadarı doğru?"
# -> Model "0" dediği tahminlerin %94'ünde, "1" dediği tahminlerin %90'ında isabetli olmuş.
# -> Yanlış alarm (False Positive) vermemek önemliyse bu skora bakılır.

# 3. RECALL (Duyarlılık / Yakalama Oranı) -> "Gerçektekilerin ne kadarını yakaladım?"
# -> Gerçekteki "0"ların %90'ını, gerçekteki "1"lerin ise %94'ünü model yakalamayı başarmış.
# -> Gözden veri kaçırmamak (örn: kanser teşhisi, dolandırıcılık tespiti) kritikse buna bakılır.

# 4. F1-SCORE (Denge Skoru: 0.92)
# -> Precision ve Recall değerlerinin dengeli bir ortalamasıdır. İkisi de yüksek olduğu için 
#    modelin çok dengeli ve stabil çalıştığını gösterir (En güvenilir metriktir).

# 5. SUPPORT (Veri Sayısı)
# -> Test setinde 151 tane '0' sınıfı, 149 tane '1' sınıfı varmış (Veri dağılımın çok dengeli).


# ==============================================================================
# CONFUSION MATRIX (KARMAŞIKLIK MATRİSİ) OKUMA REHBERİ
# ==============================================================================
# Matrisin yapısı şöyledir:
# [ [ Doğru_0 , Yanlış_0 ],
#   [ Yanlış_1 , Doğru_1 ] ]

# Senin çıktına göre yorumu:
# - 136: Gerçekte '0' olan ve modelin doğru bildiği sayı (Doğru Negatif)
# - 15 : Gerçekte '0' ama modelin yanlışlıkla '1' dediği sayı (Yanlış Alarm)
# - 9  : Gerçekte '1' ama modelin gözden kaçırıp '0' dediği sayı (Kaçırılan Veri)
# - 140: Gerçekte '1' olan ve modelin doğru bildiği sayı (Doğru Pozitif)

# ÖZET: Çapraz duran yüksek sayılar (136 ve 140) modelin başarısını, 
# diğer çaprazdaki küçük sayılar (15 ve 9) ise yaptığı hataları gösterir.
```
