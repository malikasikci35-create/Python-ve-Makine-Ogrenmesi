```python
# böyle açma kapama başa alma gibi şeylerle ugraşmamak için with komutunu kullanırız
```

```python
with open("myFile.txt") as myFile:
    myContent = myFile.read()
```

```python
myContent
```

```python
myContent # kodu istedigimiz kadar çagıralım okur üsteki gibi başa almaya gerek kalmaz
```

```python
## AŞAGIDAKİ DURUM ÇOK ÖNEMLİ DİKAT ET
```

```python
with open("myFile.txt",mode="w") as myNewFile:
    myNewFile.write("test4")  # burdaki  w write dan gelir şunu yapar içerisini siler ve ne yzarsak onu ekler
```

```python
myNewFile
```

```python
with open("myFile.txt",mode="r") as myFile2:
    myContent = myFile2.read()
```

```python
myContent # 2 tane kod yazdık birinde ekleme yaptık w ile digerinde okuduk kodu yazarken modu belitmek önemli olur
          # bu sayede istemedigimiz şeyleri yapma imkanıız ortadan kalkar
```
