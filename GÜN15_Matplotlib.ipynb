```python
numpy_arr1 = np.linspace(0,10,20) # bu kod 0 dan 20 ye kadar esit araliklarala bize degerler veriyor 
```

```python
numpy_arr1 
```

```python
numpy_arr2 = numpy_arr1 ** 3 
```

```python
numpy_arr2
```

```python
plt.plot(numpy_arr1,numpy_arr2,"g")
plt.show()
# ustel grafik
```

```python
# grafik cizgisinide baska sekilde gostere biliriz 
```

```python
plt.plot(numpy_arr1,numpy_arr2,"y*")
plt.show()
```

```python
plt.plot(numpy_arr1,numpy_arr2,"b*-")
plt.show()
```

```python
plt.plot(numpy_arr2,numpy_arr1,"b*-")
plt.show()
```

```python
# birlesik bir grafik degilde yan yana 2 tane grafik gormek istersek subplot komutunu kullanmamiz gerekir
```

```python
plt.subplot(1,2,1)
plt.plot(numpy_arr1,numpy_arr2,"r*-")
plt.subplot(1,2,2)
plt.plot(numpy_arr2,numpy_arr1,"b--")
plt.show()
```

```python
# eger eksenlerden memnun degilsek bunlari kendmizde ayarlabiliriz
```

```python
my_figure = plt.figure()
figure_axes = my_figure.add_axes([0.1,0.1,0.7,0.7])
figure_axes.plot(numpy_arr1,numpy_arr2,"g")
figure_axes.set_xlabel("X axis")
figure_axes.set_ylabel("Y axis")
figure_axes.set_title("Graph Title")
plt.show()  # burada tablomuzun yuksekligini genisligini kolonlarimizin isimlerini tek tek ayarladik  yukarida degerleri degistirince tablomuzun buyukludugude degisiyor 
```

```python
new_fig = plt.figure(dpi=100) # cozunurlugu belirliyoruz 
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_arr1,numpy_arr2 ** 2 , label = "numpy array ** 2")
new_axes.plot(numpy_arr1,numpy_arr2 ** 3 , label = "numpy array ** 3")
new_axes.legend()# tablomuzun icinde mini bir bilgilendirme tablosu gosterir 
plt.show()
```

```python
data1 = np.linspace(0,10,20)
```

```python
data1
```

```python
data2 = data1 ** 2
```

```python
plt.subplots() # subplot komutu ayni grafik uzerinden farkli bir cok grafik olusturmamuiz saglar
```

```python
my_fig, my_axes = plt.subplots()
```

```python
type(my_fig)
```

```python
type(my_axes)
```

```python
my_axes.plot(data1,data2,color = "y",alpha=0.5) # alpha kodmuz burda rengin belirligini ayarliyor 
my_axes.plot(data2,data1)
plt.show()
```

```python
(new_fig,new_axes) = plt.subplots()

new_axes.plot(data1,data1 + 2,color ="blue",alpha=1.0,linewidth=1.0)
new_axes.plot(data1,data1 + 3 ,color = "black",alpha=0.3,linewidth=4.0)# linewidth komutuda cizginin kalinligini ayarlar
new_axes.plot(data1,data1 + 4 ,color = "black",alpha=0.3,linewidth=4.0, linestyle="--") # cizgiyi ayarladik
new_axes.plot(data1,data1 + 5 ,color = "black",alpha=0.3,linewidth=4.0, marker="o",markersize = 5 , markerfacecolor="red" )# grafigimizdeki cizgiye isaret koyduk
plt.show()
```
